```python
#Exercise 15.1
#This program uses multiprocessing to create three processes. 
#Each process waits for a random time between 0 and 1 second, prints the current time, and then exits.
import threading
import time
import random
from datetime import datetime

def show_time():
    time.sleep(random.random())
    print("Time:", datetime.now())

threads = []

for _ in range(3):
    t = threading.Thread(target=show_time)
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```
