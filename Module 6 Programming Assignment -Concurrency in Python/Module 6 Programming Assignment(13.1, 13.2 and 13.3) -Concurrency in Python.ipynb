```python
#Exercise: 13.1
#This exercise demonstrates how to get the current date using datetime.
from datetime import date

# Get today's date
today = date.today()

# Convert to string (ISO format: YYYY-MM-DD)
today_string = today.isoformat()

# Write to file
with open("today.txt", "w") as file:
    file.write(today_string)

print("Saved date:", today_string)
```

```python
#Exercise 13.2
#This reads the content of today.txt and stores it in a variable.
with open("today.txt", "r") as file:
    today_string = file.read()

print("Read date:", today_string)
```

```python
#Exercise 13.3
#This converts the string back into a date object using strptime().
from datetime import datetime

parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

print("Parsed date object:", parsed_date)
```

```python
#Exercise 15.1
#This program uses multiprocessing to create three processes. 
#Each process waits for a random time between 0 and 1 second, prints the current time, and then exits.
import threading
import time
import random
from datetime import datetime

def show_time():
    time.sleep(random.random())
    print("Time:", datetime.now())

threads = []

for _ in range(3):
    t = threading.Thread(target=show_time)
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```
