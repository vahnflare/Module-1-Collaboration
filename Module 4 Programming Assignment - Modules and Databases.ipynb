```python
#16.2 - Yes, DictReader correctly handled the quotes and comma in "Eats, Shoots & Leaves". The title was treated as a single field.
import csv

with open("books.csv", newline="", encoding="utf-8") as f:
    books = list(csv.DictReader(f))

print(books)
```

```python
#16.4
import sqlite3

conn = sqlite3.connect("books.db")
cur = conn.cursor()

cur.execute("DROP TABLE IF EXISTS books")
cur.execute("""
CREATE TABLE books (
    title TEXT,
    author TEXT,
    year INTEGER
)
""")

conn.commit()
conn.close()
```

```python
#16.5
import csv
import sqlite3

conn = sqlite3.connect("books.db")
cur = conn.cursor()

with open("books2.csv", newline="", encoding="utf-8") as f:
    dr = csv.DictReader(f)
    for row in dr:
        cur.execute(
            "INSERT INTO books (title, author, year) VALUES (?, ?, ?)",
            (row["title"], row["author"], int(row["year"]))
        )

conn.commit()
conn.close()
```

```python
#16.6
import sqlite3

conn = sqlite3.connect("books.db")
cur = conn.cursor()

for (title,) in cur.execute("SELECT title FROM books ORDER BY title ASC"):
    print(title)

conn.close()
```

```python
#16.7 - The books were selected and printed in order of publication using ORDER BY year ASC.
import sqlite3

conn = sqlite3.connect("books.db")
cur = conn.cursor()

for row in cur.execute("SELECT title, author, year FROM books ORDER BY year ASC"):
    print(row)

conn.close()
```

```python
#16.8 - I used SQLAlchemy to connect to books.db and selected the title column from the books table in alphabetical order using ORDER BY title ASC.
from sqlalchemy import create_engine, text

engine = create_engine("sqlite:///books.db")

with engine.connect() as conn:
    result = conn.execute(
        text("SELECT title FROM books ORDER BY title ASC")
    )
    
    for row in result:
        print(row[0])
```
