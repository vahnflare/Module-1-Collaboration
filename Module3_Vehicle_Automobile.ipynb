```python
"""
Author: Ivanloe Manuel
Description: This program creates a simple Python application using classes and inheritance. 
A superclass named Vehicle stores the vehicle type, while a subclass named Automobile inherits from it and adds 
attributes such as year, make, model, number of doors, and roof type.
"""

class Vehicle:
    def __init__(self, vehicle_type: str):
        self.vehicle_type = vehicle_type


class Automobile(Vehicle):
    def __init__(self, year: int, make: str, model: str, doors: int, roof: str):
        super().__init__("car")  # required: store "car" into vehicle type
        self.year = year
        self.make = make
        self.model = model
        self.doors = doors
        self.roof = roof


def get_int(prompt: str, valid_values=None) -> int:
    """Get an integer from the user; optionally restrict to a set of valid values."""
    while True:
        try:
            value = int(input(prompt).strip())
            if valid_values is not None and value not in valid_values:
                print(f"Please enter one of these values: {sorted(valid_values)}")
                continue
            return value
        except ValueError:
            print("Please enter a valid number.")


def get_choice(prompt: str, choices) -> str:
    """Get a choice from the user (case-insensitive), return normalized choice."""
    normalized = {c.lower(): c for c in choices}
    while True:
        value = input(prompt).strip().lower()
        if value in normalized:
            return normalized[value]
        print(f"Please enter one of: {', '.join(choices)}")


def main():
    print("Enter car information.\n")

    year = get_int("Year (e.g., 2022): ")
    make = input("Make (e.g., Toyota): ").strip()
    model = input("Model (e.g., Corolla): ").strip()
    doors = get_int("Number of doors (2 or 4): ", valid_values={2, 4})
    roof = get_choice("Type of roof (solid or sun roof): ", choices=["solid", "sun roof"])

    car = Automobile(year=year, make=make, model=model, doors=doors, roof=roof)

    print("\nCar information:")
    print(f"  Vehicle type: {car.vehicle_type}")
    print(f"  Year: {car.year}")
    print(f"  Make: {car.make}")
    print(f"  Model: {car.model}")
    print(f"  Number of doors: {car.doors}")
    print(f"  Type of roof: {car.roof}")


main()
```
