```python
#Exercise: 4.1
#This code shows how to compare two numbers using if/elif/else to print whether 
#the guess is too low, too high, or just right.
secret = 7
guess = 5

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```

```python
#Exercise 4.2
#This exercise uses two boolean variables, small and green, to describe an object by its size and color.
small = True
green = False

if small and green:
    print("pea")
elif small and not green:
    print("cherry")
elif not small and green:
    print("watermelon")
else:
    print("pumpkin")
```
