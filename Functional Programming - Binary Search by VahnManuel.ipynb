```python
#Function Programming - Binary Search
#Test 1
def binary_search(arr, k):
    left = 0
    right = len(arr) - 1
    result = -1  # store smallest index if found

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == k:
            result = mid
            right = mid - 1  # move left to find smallest index
        elif arr[mid] < k:
            left = mid + 1
        else:
            right = mid - 1

    return result
arr1 = [1, 2, 3, 4, 5]
k = 4
print(binary_search(arr1, k1))
```
